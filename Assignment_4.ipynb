## Question 2

```python
def score_query(query, df):
    query_words = query.lower().split()
    results = []
    for index, row in df.iterrows():
        text = (row["question"]+" "+row["keywords"]).lower()
        score = 0
        for word in query_words:
            if word in text:
                score += 1
        if score > 0:
            confidence = score / len(query_words)
            results.append({
                "question":row["question"],
                "answer":row["answer"],
                "category":row["category"],
                "confidence":confidence
            })
    results.sort(key=lambda x:x["confidence"], reverse=True)
    return results
```

```python
query = input("Enter your query: ")

results = score_query(query, df)

for result in results:
    print(result)
```

## Question 6

```python
def score_query_v2(query, df):
    query_words = query.lower().split()
    results = []
    for index, row in df.iterrows():
        text = (row["question"] + " " + row["keywords"]).lower()
        score = 0
        for word in query_words:
            if word in text:
                score += 1
        if score > 0:
            confidence = score / len(query_words)
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "confidence": confidence
            })
    if not results:
        return []
        
    highest_score = max(result["confidence"] for result in results)
    best_results = [result for result in results if result["confidence"] == highest_score]

    return best_results
```

```python
print("\nTie:")
results = score_query_v2("fee", df)
for result in results:
    print("\nQuestion:", result["question"])
    print("Answer:", result["answer"])
    print("Category:", result["category"])
    print("Confidence:", result["confidence"])
```

```python
print("\nNon-Tie:")
results = score_query_v2("password", df)
for result in results:
    print("\nQuestion:", result["question"])
    print("Answer:", result["answer"])
    print("Category:", result["category"])
    print("Confidence:", result["confidence"])
```
