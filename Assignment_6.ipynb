```python
import numpy as np
```

```python
temperature=np.array([25,28,31,35,38,27,33,40])
temperature=temperature+2
print("Corrected temperature:",temperature)
F=9/5*temperature+32
print("Temperature in Fahrenheit:",F)
readings=temperature[temperature>32]
print("Readings greater than 32:",readings)
count=np.sum(temperature>32)
print("Count:",count)
```

```python
steps=np.array([[5000,6200,7100],[8000,7500,9000],[4500,5100,4800],[9000,8500,9500]])
print("Total steps:",steps.sum())
print("Mean steps:",steps.mean())
print("Maximum:",steps.max())
print("Minimum:",steps.min())
print("Total steps for each day:",steps.sum(axis=0))
print("Total steps for each user:",steps.sum(axis=1))
position=np.unravel_index(steps.argmax(),steps.shape)
print("Position of maximum:",position)
```

```python
original=np.array([1,2,3,4,5,6])
subset=original[1:5]
subset[0]=999
print("Original:",original)
print("Subset:",subset)
copied=original[1:5].copy()
copied[0]=500
print("Original:",original)
print("Copied:",copied)
arr=np.arange(1,13).reshape(3,4)
print("Matrix:")
print(arr)
print("First row:",arr[0])
print("Last row:",arr[-1])
print("Second column:",arr[:,1])
print("Rows 1-2 and columns 2-3:",arr[0:2,1:3])

flat1=arr.flatten()
flat2=arr.ravel()
print("Flatten:",flat1)
print("Ravel:",flat2)

flat2[0]=999
print("After modifying ravel:")
print("Original:",arr)
print("Ravel:",flat2)

flat1[1]=500
print("After modifying flatten:")
print("Original:",arr)
print("Flatten:",flat1)

print("Shape:",arr.shape)
print("Dimensions:",arr.ndim)
print("Size:",arr.size)
print("Data type:",arr.dtype)
```

```python
X=np.array([[6,70,3],[5,50,6],[8,80,2],[4,30,8]])
y=np.array([40,65,30,85])
print("Shape:",X.shape)
print("Dimensions:",X.ndim)
print("Transpose:")
print(X.T)
print("X.T @ X:")
print(X.T@X)
inverse=np.linalg.inv(X.T@X)
print("Inverse:")
print(inverse)
beta=inverse@X.T@y
print("Coefficients:",beta)
new_user=np.array([5,40,7])
prediction=new_user@beta
print("Predicted assistance score:",prediction)
```
